```python
import gdown
import zipfile

# stored on one of our personal drives with access permission set to "anyone with the link"
url = 'https://drive.google.com/file/d/1goEdj4cn-DT0Scf76GzHzXSwusSgSLwN/view?usp=sharing'
output_path = 'data/GTSRB.zip'
gdown.download(url, output_path, quiet=False,fuzzy=True)

with zipfile.ZipFile('data/GTSRB.zip', 'r') as zip_ref:
    zip_ref.extractall("data")
```
